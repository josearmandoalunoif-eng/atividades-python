1. Analisando temperaturas

```python
temperaturas = []
for i in range(7):
    temperaturas.append(float(input("Temperatura: ")))

soma = 0
for t in temperaturas:
    soma += t

media = soma / len(temperaturas)

acima = abaixo = iguais = 0
for t in temperaturas:
    if t > media:
        acima += 1
    elif t < media:
        abaixo += 1
    else:
        iguais += 1

print("Temperaturas:", temperaturas)
print("Média:", media)
print("Acima:", acima, "Abaixo:", abaixo, "Iguais:", iguais)
```

4. Maior sequência de crescimento

```python

numeros = []


for i in range(10):
    n = int(input("Digite um numero: "))
    numeros.append(n)


atual = 1
maior = 1


for i in range(1, 10):
    if numeros[i] > numeros[i-1]:
        atual = atual + 1
    else:
        atual = 1

    if atual > maior:
        maior = atual

print("Lista:", numeros)
print("Maior sequencia crescente:", maior)
```

7. Boletim de uma turma

```python
turma = []
for _ in range(6):
    nome = input("\nNome do aluno: ")
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    n3 = float(input("Nota 3: "))
    media = (n1 + n2 + n3) / 3
    aluno = {
        "nome": nome,
        "notas": [n1, n2, n3],
        "media": media
    }
    turma.append(aluno)
soma_todas = 0
for a in turma:
    soma_todas += a["media"]
media_turma = soma_todas / 6
maior_media = -1
melhores = []
acima_media = 0

for a in turma:
    if a["media"] > media_turma:
        acima_media += 1

    if a["media"] > maior_media:
        maior_media = a["media"]
        melhores = [a["nome"]]
    elif a["media"] == maior_media:
        melhores.append(a["nome"])

print("\n===== BOLETIM =====")
for aluno in turma:
    print(f"{aluno['nome']} - Média: {aluno['media']:.1f}")

print(f"\nMédia da turma: {media_turma:.1f}")
print(f"Alunos acima da média: {acima_media}")
print(f"Melhor(es) aluno(s): {melhores}")
```

8. Controle de estoque com movimentações

```python

produtos = []
for i in range(5):
    codigo = input(f"Código do produto {i+1}: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade em estoque: "))
    produtos.append({"codigo": codigo, "nome": nome, "preco": preco, "quantidade": quantidade})

num_mov = int(input("Quantas movimentações deseja realizar? "))


for i in range(num_mov):
    codigo = input("Código do produto para movimentação: ")
    tipo = input("Tipo (E=entrada, S=saída): ").upper()
    qtd = int(input("Quantidade: "))

    produto = None
    for p in produtos:
        if p["codigo"] == codigo:
            produto = p
            break

    if produto:
        if tipo == "E":
            produto["quantidade"] += qtd
            print(f"Entrada registrada: {produto['nome']} +{qtd}")
        elif tipo == "S":
            if produto["quantidade"] >= qtd:
                produto["quantidade"] -= qtd
                print(f"Saída registrada: {produto['nome']} -{qtd}")
            else:
                print("Saída recusada: estoque insuficiente.")
        else:
            print("Tipo inválido.")
    else:
        print("Produto não encontrado.")


valor_total = 0
print("\n--- ESTOQUE ATUALIZADO ---")
for p in produtos:
    print(p)
    valor_total += p["quantidade"] * p["preco"]

print("Valor financeiro total do estoque:", valor_total)
```

10. Agenda de contatos sem duplicidade

```python
contatos = []
num_contatos = int(input("Quantos contatos deseja cadastrar? "))

for i in range(num_contatos):
    nome = input("Nome: ")

    # Verifica se já existe
    existe = False
    for c in contatos:
        if c["nome"] == nome:
            existe = True
            break

    if existe:
        print("Contato já existe!")
    else:
        telefone = input("Telefone: ")
        contatos.append({"nome": nome, "telefone": telefone})

consulta = input("Consultar nome: ")
encontrado = False
for c in contatos:
    if c["nome"] == consulta:
        print("Telefone:", c["telefone"])
        encontrado = True
        break

if not encontrado:
    print("Contato não encontrado.")
```

11. Pesquisa de votos

```python

votos = {1:0, 2:0, 3:0, 4:0}
num_votos = int(input("Quantos votos serão registrados? "))

for i in range(num_votos):
    voto = int(input("Digite seu voto (1-4): "))
    if voto in votos:
        votos[voto] += 1
    else:
        print("Código inválido!")

total = sum(votos.values())

print("Resultado da votação:")
for opcao, qtd in votos.items():
    perc = (qtd / total * 100) if total > 0 else 0
    print(f"Opção {opcao}: {qtd} votos ({perc:.2f}%)")

maior = 0
vencedores = []
for opcao, qtd in votos.items():
    if qtd > maior:
        maior = qtd
        vencedores = [opcao]
    elif qtd == maior:
        vencedores.append(opcao)

print("Opção(ões) vencedora(s):", vencedores)
```

12. Histórico de movimentações

```python
estoque = {"A": 100, "B": 50, "C": 80}
movimentacoes = []

num_mov = int(input("Quantas movimentações deseja realizar? "))

for i in range(num_mov):
    codigo = input("Código do produto: ")
    tipo = input("Tipo (E=entrada, S=saída): ").upper()
    qtd = int(input("Quantidade: "))

    if tipo == "E":
        estoque[codigo] = estoque.get(codigo, 0) + qtd
    elif tipo == "S":
        if estoque.get(codigo, 0) >= qtd:
            estoque[codigo] -= qtd
        else:
            print("Saída recusada: estoque insuficiente.")
    else:
        print("Tipo inválido.")

    movimentacoes.append((codigo, tipo, qtd))

print("Estoque final:", estoque)
print("Histórico de movimentações:")
for m in movimentacoes:
    print(m)
```

13. Análise de vendas por vendedor

```python
vendedores = []

for i in range(5):
    nome = input(f"Nome do vendedor {i+1}: ")
    vendas = []
    for j in range(4):
        valor = float(input(f"Valor da venda {j+1}: "))
        vendas.append(valor)
    vendedores.append({"nome": nome, "vendas": vendas})

totais = []
total_geral = 0
for v in vendedores:
    total_vendedor = 0
    for venda in v["vendas"]:
        total_vendedor += venda
    v["total"] = total_vendedor
    totais.append(total_vendedor)
    total_geral += total_vendedor

media = total_geral / len(vendedores)
maior = 0
melhores = []
for v in vendedores:
    if v["total"] > maior:
        maior = v["total"]
        melhores = [v["nome"]]
    elif v["total"] == maior:
        melhores.append(v["nome"])

acima_media = sum(1 for v in vendedores if v["total"] > media)

print("\n--- RELATÓRIO ---")
print("Total geral:", total_geral)
for v in vendedores:
    print(v["nome"], "vendeu", v["total"])
print("Vendedor(es) com maior total:", melhores)
print("Quantidade acima da média:", acima_media)
```

14. Catálogo e pedidos

```python

produtos = []
for i in range(5):
    codigo = input(f"Código do produto {i+1}: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    estoque = int(input("Estoque inicial: "))
    produtos.append({"codigo": codigo, "nome": nome, "preco": preco, "estoque": estoque})

pedidos = []
for i in range(6):
    codigo = input(f"Código do produto do pedido {i+1}: ")
    quantidade = int(input("Quantidade: "))
    pedidos.append((codigo, quantidade))
valor_total = 0
for codigo, qtd in pedidos:
    produto = None
    for p in produtos:
        if p["codigo"] == codigo:
            produto = p
            break

    if produto and produto["estoque"] >= qtd:
        produto["estoque"] -= qtd
        valor_total += produto["preco"] * qtd
        print(f"Pedido atendido: {produto['nome']} x{qtd}")
    else:
        print(f"Pedido recusado: código {codigo}, quantidade {qtd}")

print("\nValor total dos pedidos atendidos:", valor_total)
```

```python
produtos = []
for i in range(5):
    codigo = input(f"Código do produto {i+1}: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade em estoque: "))
    produtos.append({"codigo": codigo, "nome": nome, "preco": preco, "estoque": quantidade, "vendido": 0, "valor_vendido": 0})

historico = []
num_vendas = int(input("Quantas vendas deseja registrar? "))

for i in range(num_vendas):
    codigo = input("Código do produto: ")
    qtd = int(input("Quantidade vendida: "))

    produto = None
    for p in produtos:
        if p["codigo"] == codigo:
            produto = p
            break

    if produto and produto["estoque"] >= qtd:
        produto["estoque"] -= qtd
        produto["vendido"] += qtd
        valor = produto["preco"] * qtd
        produto["valor_vendido"] += valor
        historico.append((codigo, qtd, valor))
        print(f"Venda registrada: {produto['nome']} x{qtd}")
    else:
        print("Venda recusada (produto inexistente ou estoque insuficiente).")

total_unidades = sum(p["vendido"] for p in produtos)
valor_total = sum(p["valor_vendido"] for p in produtos)

maior_valor = 0
melhores = []
for p in produtos:
    if p["valor_vendido"] > maior_valor:
        maior_valor = p["valor_vendido"]
        melhores = [p["nome"]]
    elif p["valor_vendido"] == maior_valor:
        melhores.append(p["nome"])

sem_venda = [p["nome"] for p in produtos if p["vendido"] == 0]
resumo = (len(historico), total_unidades, valor_total)

print("\n--- RELATÓRIO FINAL ---")
for p in produtos:
    print(p)
print("Unidades vendidas:", total_unidades)
print("Valor total:", valor_total)
print("Produto(s) de maior valor vendido:", melhores)
print("Produtos sem venda:", sem_venda)
print("Histórico de vendas:", historico)
print("Resumo:", resumo)
```
