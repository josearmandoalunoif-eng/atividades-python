Resolução comentada · Parte 1

```python
notas = []
for i in range(5):
    while True:
        try:
            nota = float(input(f"Digite a {i+1}ª nota: "))
            if not 0 <= nota <= 10:
                print("A nota deve estar entre 0 e 10.")
                continue
            notas.append(nota)
            break
        except ValueError:
            print("Entrada inválida. Digite um número.")

soma = 0
for nota in notas:
    soma += nota
media = soma / len(notas)

acima_ou_igual = 0
for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

Resolução comentada · Parte 2

```python
def ler_numero(msg):
    while True:
        try:
            return float(input(msg))
        except ValueError:
            print("Entrada inválida. Digite um número.")

x = ler_numero("Digite x: ")
y = ler_numero("Digite y: ")
ponto = (x, y)

if x == 0 and y == 0:
    pos = "Origem"
elif x == 0:
    pos = "Sobre o eixo Y"
elif y == 0:
    pos = "Sobre o eixo X"
elif x > 0 and y > 0:
    pos = "1º Quadrante"
elif x < 0 and y > 0:
    pos = "2º Quadrante"
elif x < 0 and y < 0:
    pos = "3º Quadrante"
else:
    pos = "4º Quadrante"

print(f"Ponto: {ponto} -> {pos}")
```

Questão 3 · Cadastro simples de produto

```python
nome = input("Nome do produto: ")
while nome.strip() == "":
    print("Nome não pode ser vazio.")
    nome = input("Nome do produto: ")

while True:
    try:
        preco = float(input("Preço: "))
        if preco < 0:
            print("Preço deve ser >= 0.")
            continue
        break
    except ValueError:
        print("Entrada inválida. Digite um número.")

while True:
    try:
        qtd = int(input("Quantidade: "))
        if qtd < 0:
            print("Quantidade deve ser >= 0.")
            continue
        break
    except ValueError:
        print("Entrada inválida. Digite um número inteiro.")

produto = {"nome": nome, "preco": preco, "quantidade": qtd}
total = preco * qtd

print(f"Cadastro: {produto}")
print(f"Valor total em estoque: R$ {total:.2f}")
```

Questão 4 · Lista de números: pares e ímpares

```python
numeros = []
for i in range(10):
    while True:
        try:
            n = int(input(f"Digite o {i+1}º número: "))
            numeros.append(n)
            break
        except ValueError:
            print("Entrada inválida. Digite um número inteiro.")

pares = []
impares = []
for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print(f"Todos: {numeros}")
print(f"Pares: {pares}")
print(f"Ímpares: {impares}")
```

Questão 5 · Maior e menor da lista

```python
numeros = []
for i in range(8):
    while True:
        try:
            n = int(input(f"Digite o {i+1}º número: "))
            numeros.append(n)
            break
        except ValueError:
            print("Entrada inválida.")

maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

for i, n in enumerate(numeros):
    if n > maior:
        maior = n
        pos_maior = i
    if n < menor:
        menor = n
        pos_menor = i

print(f"Lista: {numeros}")
print(f"Maior: {maior} na posição {pos_maior}")
print(f"Menor: {menor} na posição {pos_menor}")
```

Questão 6 · Nomes sem repetição

```python
nomes = []
while True:
    nome = input("Digite um nome (FIM para parar): ").strip()
    if nome.upper() == "FIM":
        break
    if nome == "":
        print("Nome vazio, digite novamente.")
        continue
    if nome not in nomes:
        nomes.append(nome)
    else:
        print("Nome já cadastrado, não será repetido.")

print(f"Nomes: {nomes}")
print(f"Quantidade de nomes diferentes: {len(nomes)}")
```

Questão 8 · Estatísticas de uma tupla

```python
valores = []
for i in range(6):
    while True:
        try:
            n = int(input(f"Digite o {i+1}º número: "))
            valores.append(n)
            break
        except ValueError:
            print("Entrada inválida.")

tupla = tuple(valores)
qtd_cinco = tupla.count(5)

if 5 in tupla:
    pos = tupla.index(5)  # primeira ocorrência
else:
    pos = "não existe"

pares = 0
for n in tupla:
    if n % 2 == 0:
        pares += 1

print(f"Tupla: {tupla}")
print(f"Número 5 apareceu: {qtd_cinco} vezes")
print(f"Primeira ocorrência de 5: {pos}")
print(f"Quantidade de pares: {pares}")
```

Questão 9 · Dicionário de aluno

```python
nome = input("Nome: ")
turma = input("Turma: ")

def ler_nota(msg):
    while True:
        try:
            n = float(input(msg))
            if 0 <= n <= 10:
                return n
            print("Nota deve estar entre 0 e 10.")
        except ValueError:
            print("Entrada inválida.")

nota1 = ler_nota("Nota 1: ")
nota2 = ler_nota("Nota 2: ")

aluno = {"nome": nome, "turma": turma, "nota1": nota1, "nota2": nota2}
media = (nota1 + nota2) / 2
aluno["media"] = media

if media >= 7:
    aluno["situacao"] = "aprovado"
elif media >= 5:
    aluno["situacao"] = "recuperação"
else:
    aluno["situacao"] = "reprovado"

print(aluno)
```

Questão 11 · Frequência das palavras

```python
frequencia = {}
for i in range(8):
    palavra = input(f"Digite a {i+1}ª palavra: ").strip().lower()
    if palavra == "":
        print("Palavra vazia, digite novamente.")
        continue
    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

for palavra, qtd in frequencia.items():
    print(f"{palavra}: {qtd} vez(es)")
```

Questão 12 · Estoque e atualização

```python
estoque = {"arroz": 20, "feijão": 15, "açúcar": 10, "café": 8}
print(f"Estoque inicial: {estoque}")

produto = input("Produto: ").strip().lower()

while True:
    try:
        vendida = int(input("Quantidade vendida: "))
        if vendida < 0:
            print("Quantidade deve ser >= 0.")
            continue
        break
    except ValueError:
        print("Entrada inválida.")

if produto not in estoque:
    print("Produto não encontrado.")
elif vendida > estoque[produto]:
    print("Estoque insuficiente.")
else:
    estoque[produto] -= vendida
    print("Estoque atualizado!")

print(f"Estoque final: {estoque}")
```

Questão 13 · Cadastro de vários alunos

```python
alunos = []
for i in range(5):
    print(f"\n--- Aluno {i+1} ---")
    nome = input("Nome: ")
    while True:
        try:
            n1 = float(input("Nota 1: "))
            n2 = float(input("Nota 2: "))
            if not (0 <= n1 <= 10 and 0 <= n2 <= 10):
                print("Notas devem estar entre 0 e 10.")
                continue
            break
        except ValueError:
            print("Entrada inválida.")

    media = (n1 + n2) / 2
    aluno = {"nome": nome, "notas": (n1, n2), "media": media}
    alunos.append(aluno)

print("\n--- Resultado ---")
for a in alunos:
    print(f"{a['nome']} - Média: {a['media']:.1f}")
```

Questão 14 · Catálogo com código imutável

```python
catalogo = []
for i in range(5):
    print(f"\nProduto {i+1}")
    while True:
        try:
            codigo = int(input("Código: "))
            break
        except ValueError:
            print("Código deve ser número inteiro.")
    nome = input("Nome: ")
    while True:
        try:
            preco = float(input("Preço: "))
            break
        except ValueError:
            print("Preço inválido.")

    produto = (codigo, nome, preco)
    catalogo.append(produto)

while True:
    try:
        busca = int(input("\nDigite o código para buscar: "))
        break
    except ValueError:
        print("Entrada inválida.")

encontrado = False
for codigo, nome, preco in catalogo:
    if codigo == busca:
        print(f"Encontrado: {nome} - R$ {preco:.2f}")
        encontrado = True
        break

if not encontrado:
    print("Código não encontrado.")
```

Questão 15 · Mini sistema de estoque

```python
estoque = []
for i in range(5):
    print(f"\nItem {i+1}")
    while True:
        try:
            codigo = int(input("Código: "))
            qtd = int(input("Quantidade: "))
            preco = float(input("Preço: "))
            if qtd < 0 or preco < 0:
                print("Valores devem ser >= 0.")
                continue
            break
        except ValueError:
            print("Entrada inválida. Digite números.")
    nome = input("Nome: ")
    item = {"codigo": codigo, "nome": nome, "quantidade": qtd, "preco": preco}
    estoque.append(item)

print("\n--- Todos os itens ---")
for item in estoque:
    print(item)

total = 0
for item in estoque:
    total += item["quantidade"] * item["preco"]
print(f"\nValor total do estoque: R$ {total:.2f}")

maior_item = estoque[0]
maior_valor = estoque[0]["quantidade"] * estoque[0]["preco"]
for item in estoque:
    valor = item["quantidade"] * item["preco"]
    if valor > maior_valor:
        maior_valor = valor
        maior_item = item
print(f"Maior valor em estoque: {maior_item['nome']} - R$ {maior_valor:.2f}")

resumo = (len(estoque), total)
print(f"Resumo (qtd_itens, valor_total): {resumo}")
```
