Questão 1 - Maior e menor entre vários valores

```python
def ler_inteiro():
    while True:
        try:
            return int(input("Digite um número: "))
        except ValueError:
            print("Entrada inválida. Digite um inteiro.")

primeiro = ler_inteiro()
maior = menor = primeiro

for i in range(5):
    numero = ler_inteiro()
    if numero > maior:
        maior = numero
    if numero < menor:
        menor = numero

print("Maior:", maior)
print("Menor:", menor)
```

Questão 2 - Estatísticas de uma turma

```python
soma = 0
aprovados = 0
abaixo = 0

for i in range(5):
    while True:
        try:
            nota = float(input(f"Nota do aluno {i+1}: "))
            if not 0 <= nota <= 10:
                print("Nota deve estar entre 0 e 10.")
                continue
            break
        except ValueError:
            print("Digite uma nota numérica válida.")
    soma += nota
    if nota >= 7:
        aprovados += 1
    else:
        abaixo += 1

print("Média:", soma / 5)
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

Questão 5 - Máquina de venda de ingresso

```python
PRECO = 25
total_inserido = 0

while total_inserido < PRECO:
    try:
        valor = int(input(f"Insira um valor (Falta R$ {PRECO - total_inserido}): "))
        if valor <= 0:
            print("O depósito deve ser inteiro positivo.")
            continue
        total_inserido += valor
    except ValueError:
        print("Texto inválido não soma. Digite apenas números inteiros.")

troco = total_inserido - PRECO
print(f"Total inserido: R$ {total_inserido}")
print(f"Troco: R$ {troco}")
```

Questão 9 - Controle de saldo

```python
saldo = 500.00

for i in range(6):
    print(f"\nOperação {i+1} - Saldo: R$ {saldo:.2f}")
    while True:
        try:
            tipo = int(input("Tipo (1=depósito, 2=saque): "))
            if tipo not in [1, 2]:
                print("Código inválido. Use 1 ou 2.")
                continue
            break
        except ValueError:
            print("Entrada inválida.")

    while True:
        try:
            valor = float(input("Valor: "))
            if valor <= 0:
                print("Valor deve ser positivo.")
                continue
            break
        except ValueError:
            print("Entrada inválida.")

    if tipo == 1:
        saldo += valor
    else:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente.")

print(f"\nSaldo final: R$ {saldo:.2f}")
```

Questão 11 - Número primo

```python
while True:
    try:
        n = int(input("Digite um inteiro maior que 1: "))
        if n <= 1:
            print("O número deve ser maior que 1.")
            continue
        break
    except ValueError:
        print("Entrada inválida.")

eh_primo = True
if n > 2:
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            eh_primo = False
            break

if eh_primo:
    print(f"{n} é primo.")
else:
    print(f"{n} não é primo.")
```

Questão 12 - Divisores de um número

```python
while True:
    try:
        n = int(input("Digite N positivo: "))
        if n <= 0:
            print("N deve ser > 0.")
            continue
        break
    except ValueError:
        print("Entrada inválida.")

divisores = []
for i in range(1, n + 1):
    if n % i == 0:
        divisores.append(i)

print(f"Divisores de {n}: {divisores}")
print(f"Quantidade: {len(divisores)}")
```

Questão 13 - Números primos em um intervalo

```python
while True:
    try:
        a = int(input("Digite A positivo: "))
        if a <= 0:
            print("A deve ser > 0.")
            continue
        break
    except ValueError:
        print("Entrada inválida.")

while True:
    try:
        b = int(input("Digite B positivo (B > A): "))
        if b <= 0 or b <= a:
            print("B deve ser > 0 e maior que A.")
            continue
        break
    except ValueError:
        print("Entrada inválida.")

primos = []
for num in range(a, b + 1):
    if num <= 1:
        continue
    eh_primo = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            eh_primo = False
            break
    if eh_primo:
        primos.append(num)

print(f"Primos entre {a} e {b}: {primos}")
print(f"Total encontrados: {len(primos)}")
```

Questão 14 - Jogo com limite de tentativas

```python
SECRETO = 37
tentativas = 0

while tentativas < 5:
    try:
        palpite = int(input(f"Tentativa {tentativas+1}/5 - Digite seu palpite: "))
    except ValueError:
        print("Texto inválido não gasta tentativa. Digite um número.")
        continue

    tentativas += 1

    if palpite == SECRETO:
        print(f"Parabéns! Você acertou em {tentativas} tentativas.")
        break
    else:
        if tentativas == 5:
            print(f"Suas tentativas acabaram. O número secreto era {SECRETO}.")
        elif palpite < SECRETO:
            print("O secreto é MAIOR que seu palpite.")
        else:
            print("O secreto é MENOR que seu palpite.")
```

Questão 15 - Desafio final - caixa de atendimento

```python
saldo = 200.0
qtd_entrada = 0
qtd_saida_ok = 0
qtd_saida_recusada = 0
total_entrou = 0
total_saiu = 0

while True:
    while True:
        try:
            codigo = int(input("Codigo (0 sair, 1 entrada, 2 retirada): "))
            if codigo not in [0, 1, 2]:
                print("Codigo invalido, use 0, 1 ou 2")
                continue
            break
        except ValueError:
            print("Digite so numero")

    if codigo == 0:
        break
    while True:
        try:
            valor = float(input("Valor: "))
            if valor <= 0:
                print("Valor tem que ser maior que zero")
                continue
            break
        except ValueError:
            print("Digite so numero")

    if codigo == 1:
        saldo = saldo + valor
        qtd_entrada = qtd_entrada + 1
        total_entrou = total_entrou + valor

    if codigo == 2:
        if valor <= saldo:
            saldo = saldo - valor
            qtd_saida_ok = qtd_saida_ok + 1
            total_saiu = total_saiu + valor
        else:
            print("Saldo insuficiente")
            qtd_saida_recusada = qtd_saida_recusada + 1

print(f"Saldo final: {saldo}")
print(f"Qtd entradas: {qtd_entrada}")
print(f"Qtd retiradas feitas: {qtd_saida_ok}")
print(f"Qtd retiradas recusadas: {qtd_saida_recusada}")
print(f"Total entrou: {total_entrou}")
print(f"Total saiu: {total_saiu}")
```
