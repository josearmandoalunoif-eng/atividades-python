#4. Estacionamento

```python
horas = int(input("Quantas horas ficou estacionado: "))

if horas <= 2:
    valor = 8.00
elif horas <= 5:
    valor = 8.00 + (horas - 2) * 3.00
else:
    valor = 17.00 + (horas - 5) * 2.00

print(f"Valor a pagar: R$ {valor:.2f}")
```

#5. Máquina de venda de ingresso

```python
preco = 25
total = 0
while total < preco:
    valor = int(input("Insira valor: "))
    total = total + valor
troco = total - preco
print("Total inserido:", total)
print("Troco:", troco)
```

#6. Maior valor e sua posição

```python
maior = None
posicao_maior = 0

for i in range(1, 9):
    numero = int(input(f"Digite o {i}º número: "))

    if maior is None or numero > maior:
        maior = numero
        posicao_maior = i

print("Maior valor:", maior)
print("Primeira posição:", posicao_maior)
```

#7. Segundo maior valor

```python
n1 = int(input("1.º número: "))
n2 = int(input("2.º número: "))

if n1 > n2:
    maior = n1
    segundo = n2
else:
    maior = n2
    segundo = n1

for _ in range(6):
    num = int(input("Próximo número: "))
    if num > maior:
        segundo = maior
        maior = num
    elif num > segundo:
        segundo = num

print("Maior:", maior)
print("Segundo maior:", segundo)
```

#8. Detectando repetição consecutiva

```python
anterior = None
repetiu = False

for i in range(8):
    numero = int(input("Digite um número: "))

    if anterior is not None and numero == anterior:
        repetiu = True

    anterior = numero

if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

#9. Controle de saldo

```python
saldo = 500

for i in range(6):
    tipo = int(input("Digite 1 para depósito ou 2 para saque: "))
    valor = float(input("Digite o valor: R$ "))

    if tipo == 1:
        saldo += valor

    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo final: R$ {saldo:.2f}")
```

#10. Caixa eletrônico simplificado

```python
valor = int(input("Digite o valor: "))

cedulas = [100, 50, 20, 10, 5, 2, 1]

for cedula in cedulas:
    quantidade = valor // cedula
    valor = valor % cedula

    if quantidade > 0:
        print(f"{quantidade} cédula(s) de R$ {cedula}")
```

#11. Número primo

```python
numero = int(input("Digite um número maior que 1: "))

primo = True

for i in range(2, numero):
    if numero % i == 0:
        primo = False
        break

if primo:
    print("O número é primo.")
else:
    print("O número não é primo.")
```

#12. Divisores de um número

```python
n = int(input("Digite um número positivo: "))

quantidade = 0

print("Divisores:")

for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        quantidade += 1

print("Quantidade de divisores:", quantidade)
```

#13. Números primos em um intervalo

```python
a = int(input("Digite A: "))
b = int(input("Digite B: "))

quantidade = 0

print("Números primos:")

for numero in range(a, b + 1):

    if numero > 1:
        primo = True

        for divisor in range(2, numero):
            if numero % divisor == 0:
                primo = False
                break

        if primo:
            print(numero)
            quantidade += 1

print("Quantidade de primos:", quantidade)
```

#14. Jogo com limite de tentativas

```python
secreto = 37

for i in range(5):
    palpite = int(input("Digite seu palpite: "))

    if palpite == secreto:
        print("Acertou!")
        break
    elif palpite < secreto:
        print("O número é maior.")
    else:
        print("O número é menor.")
else:
    print("Fim das tentativas.")
```

#15. Desafio final - caixa de atendimento

```python
saldo = 200
entradas = 0
saidas_ok = 0
saidas_nao = 0
total_entrou = 0
total_saiu = 0
codigo = int(input("Codigo 1=entrada 2=saida 0=sair: "))
while codigo != 0:
    valor = int(input("Valor: "))
    if codigo == 1:
        saldo = saldo + valor
        total_entrou = total_entrou + valor
        entradas = entradas + 1
        print("Entrada ok")
    if codigo == 2:
        if valor <= saldo:
            saldo = saldo - valor
            total_saiu = total_saiu + valor
            saidas_ok = saidas_ok + 1
            print("Saida ok")
        else:
            print("Nao tem saldo")
            saidas_nao = saidas_nao + 1
    if codigo != 1 and codigo != 2 and codigo != 0:
        print("Codigo errado")
    codigo = int(input("Codigo 1=entrada 2=saida 0=sair: "))
print("Saldo final:", saldo)
print("Qtd entradas:", entradas)
print("Qtd saidas ok:", saidas_ok)
print("Qtd saidas recusadas:", saidas_nao)
print("Total entrou:", total_entrou)
print("Total saiu:", total_saiu)
```
